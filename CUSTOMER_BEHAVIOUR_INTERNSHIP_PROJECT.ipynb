Look at column name uppercase lowercase mixed and spaces

A good practice to put column names in snake casing with all lowercase and underscore


```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(" ","_")
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})#im replacing purchase_amount_(usd) to purchase_amount
```

```python
df.columns
```

```python
#create a column for age_group
labels=['Young Adult','Adult','Middle-Aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)#Q card it splits the age into four equal sizes groups based on the data distribution and assign labels be defined.
```

```python
df[['age','age_group']].head(10)#Original ages alongside the new age group column, now it is easier to analyze pattern across different age category which can be really useful for marketing, understanding the customer behavior.
```

So right now, the dataset has a column called frequency of purchases, which tells us how often a customer shops, but it's in text form, like weekly, monthly, or quarterly. And you know that working with text makes analysis harder, especially if you want to calculate metrics or compare customers. So what we'll do is we'll convert these textual frequencies into numbers, basically the number of days between purchases. And this will help us make our analysis much easier.

```python
#create column purchase_frequency_days
frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

Alongside the new numeric column, now instead of just seeing weekly or monthly, you can see the exact number of days between purchases for each customer. And believe me, this is super useful for analysis.
